## POST 返回 200 之后

本课程开发时真实发生过：调用 GitCode API 创建 PR，接口返回 200，看似成功。但回查才发现，单查接口返回的状态枚举与创建响应不一致（open vs opened），字段对不上就不算完成。

**返回成功 ≠ 任务完成**，这是协作提交环节最重要的一条纪律，也是本节的起点。算子写对了（02.01 节）、跑快了（02.02 节），最后一关是进仓库；这一关规则明确、可验证、低创造性，恰好是最适合交给 Agent 的部分，也是红线最密集的部分。

## 1. 为什么最后一公里最适合自动化

提交 PR/Issue 这类协作工作有三个特征：**规则明确**（约定式提交、模板字段，对错可判定）、**可验证**（API 回查）、**低创造性**（内容应从代码 diff **推导**，而非自由创作）。三者齐备的任务正是 Agent 的主场，cannbot-skills 仓 `infra/` 域的协作 Skill 家族（gitcode-toolkit / pr-handler / issue-gen / issue-handler）就是按这个判断设计的。

**红线是安全工程**，每条对应一类真实事故：
